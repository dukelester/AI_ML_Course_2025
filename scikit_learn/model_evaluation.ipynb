# Model Evaluation

* r^2 ( r-squared) coefficient of determination 
* Mean Absolute error (MAE)
* Mean Squared Error(MSE)

```python
print('Model EValuation')
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
# Load the data

heart_disease = pd.read_csv('data/heart.csv')
heart_disease.head(4)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(42)

X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# fit the model
model = RandomForestRegressor()
model.fit(X_train, y_train)
```

```python
# Score the model

model.score(X_test, y_test)
```

```python
# Model Evaluation 1. r2_score
from sklearn.metrics import r2_score

# Fill an array with the y_test mean
y_test_mean = np.full(len(y_test), y_test.mean())

y_test_mean
```

```python

r2_score(y_true=y_test, y_pred=y_test_mean)
```

```python
r2_score(y_true=y_test, y_pred=y_test)
```

```python
# MAE (Mean Absolute Error)

from sklearn.metrics import mean_absolute_error

y_preds = model.predict(X_test)
mae = mean_absolute_error(y_test, y_preds)
mae
```

```python
y_preds
```

```python
result_data = pd.DataFrame(data = {
    "actual values": y_test,
    "predicted values": y_preds
    })

result_data['differences'] = result_data['predicted values'] - result_data['actual values']


result_data
```

```python
result_data['differences'].mean()
```

```python
np.abs(result_data['differences']).mean()
```

```python
result_data['absolute difference'] = np.abs(result_data['differences'])
result_data.head()
```

```python
result_data['absolute difference'].mean()  # Mean absolute error
```

```python
# Evaluating the model with MSE(Mean squared Error)
# The square of the errors between the actual and predicted values


from sklearn.metrics import mean_squared_error
y_preds = model.predict(X_test)
mse = mean_squared_error(y_test, y_preds)
mse
```

```python
result_data['squared differences'] = np.square(result_data['differences'])
result_data.head()
```

```python
result_data['squared differences'].mean()
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
```

```python
np.random.seed(42)

X = heart_disease.drop('target', axis=1)
y = heart_disease['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()

model.fit(X_train, y_train)

y_preds = model.predict(X_test)

#Evaluate and score the model

print(f'The Accuracy Score is {accuracy_score(y_test, y_preds) *100: .2f}')
print(f'The Precision Score is {precision_score(y_test, y_preds)}')
print(f'The Recall Score is {recall_score(y_test, y_preds)}')
print(f'The F1 Score is {f1_score(y_test, y_preds)}')
```

```python
# Evaluation
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.ensemble import RandomForestRegressor

regressor_model = RandomForestRegressor()
regressor_model.fit(X_train, y_train)

y_preds = regressor_model.predict(X_test)

print(f'R2 score {r2_score(y_test, y_preds)}')
print(f'The Mean absolute Error score {mean_absolute_error(y_test, y_preds)}')
print(f'The Mean Squared Error {mean_squared_error(y_test, y_preds)}')

```

```python

grid = {'n_estimators': [205, 820]}
grid_search_cv = GridSearchCV(estimator=regressor_model,
                            param_grid=grid,
                            n_jobs=10, cv=5, verbose=5)

grid_search_cv.fit(X_train, y_train)
y_pred = grid_search_cv.predict(X_test)
y_pred
```

```python
# Save the model

import pickle

pickle.dump(grid_search_cv, open('models/gs_random_forest_model.pkl', 'wb'))
```
